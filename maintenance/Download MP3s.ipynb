```python

import requests
from bs4 import BeautifulSoup
import os

# Function to download MP3 files
def download_mp3(mp3_links):
    for mp3_link in mp3_links:
        file_name = os.path.basename(mp3_link)
        print(f"Downloading {file_name}...")
        response = requests.get(mp3_link)
        with open(file_name, 'wb') as file:
            file.write(response.content)
    print("Download complete!")
```

```python

# Base URL of the site
domain_name = 'naturallawinstitute.com'
base_url = f"https://{domain_name}"

# Start crawling
response = requests.get(base_url)
soup = BeautifulSoup(response.text, 'html.parser')
mp3_links = []
action_links = []
for link in soup.find_all('a', href=True):
    href = link['href']
    if href.endswith('.mp3'):
        if href.startswith('http'):
            mp3_links.append(href)
        else:
            mp3_links.append(base_url + href)
    else:
        if href.startswith('http'):
            if domain_name in href:
                action_links.append(href)
        else:
            action_links.append(base_url + href)
```
