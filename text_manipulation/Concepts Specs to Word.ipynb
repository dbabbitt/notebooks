```python

%pprint
import pypandoc
import os.path as osp, os as os
import shutil
import subprocess

specs_folder = r'C:\Users\daveb\OneDrive\Documents\Projects\Concept Specs'
outputfile_path = osp.join(specs_folder, 'pet-management-improves-thanksgiving-vacation.docx')
word_path = r"C:\Program Files\Microsoft Office\root\Office16\WINWORD.EXE"
file_prefix = 'sermonaudio-hosting-improves-church-media-search'
migration_markdown_path = osp.join(specs_folder, f'{file_prefix}.md')
migration_word_path = osp.join(specs_folder, f'{file_prefix}.docx')
downloads_folder = r'C:\Users\daveb\Downloads'
left_path = osp.join(downloads_folder, 'left.txt')
right_path = osp.join(downloads_folder, f'right.txt')
```

```python

# Convert the Thanksgiving Vacation Strategy to Microsoft Word
_ = pypandoc.convert_file(
    source_file=osp.join(specs_folder, 'pet-management-improves-thanksgiving-vacation.md'),
    to='docx',
    format='markdown',
    outputfile=outputfile_path,
    verify_format=True,
    sandbox=False,
    cworkdir=None,
    sort_files=True,
)
```

```python

# Open in Microsoft Word
_ = subprocess.Popen([word_path, outputfile_path])
```

```python

# Convert the Library Migration Strategy to Microsoft Word
_ = pypandoc.convert_file(
    source_file=migration_markdown_path,
    to='docx',
    format='markdown',
    outputfile=migration_word_path,
    verify_format=True,
    sandbox=False,
    cworkdir=None,
    sort_files=True,
)
```

```python

# Compare the Microsoft Word version of the Library Migration Strategy to the original
pypandoc.convert_file(
    source_file=migration_word_path,
    to='markdown',
    format='docx',
    outputfile=right_path,
    verify_format=True,
    sandbox=False,
    cworkdir=None,
    sort_files=True,
)
_ = shutil.copy2(migration_markdown_path, left_path)
```

```python

# Open in Compare It!
diff_path = r"C:\Program Files (x86)\Compare It!\wincmp3.exe"
_ = subprocess.Popen([diff_path, left_path, right_path])
```
